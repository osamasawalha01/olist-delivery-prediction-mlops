```python
import pandas as pd
from sklearn.model_selection import train_test_split

labeled_table = pd.read_parquet(
    "../artifacts/02_labeled_orders.parquet"
)

print("Rows:", len(labeled_table))
print("Columns:", len(labeled_table.columns))
```

```python
print("Order purchase date range:")
print(
    labeled_table["order_purchase_timestamp"].min(),
    "to",
    labeled_table["order_purchase_timestamp"].max()
)

print("\nOverall label distribution:")
print(
    labeled_table["late_delivery"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
train_df, temp_df = train_test_split(
    labeled_table,
    test_size=0.30,
    random_state=42,
    stratify=labeled_table["late_delivery"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["late_delivery"]
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
for name, df in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    print(f"\n{name}")
    print(
        df["late_delivery"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )
```

```python
train_df.to_parquet(
    "../artifacts/03_train.parquet",
    index=False
)

val_df.to_parquet(
    "../artifacts/03_validation.parquet",
    index=False
)

test_df.to_parquet(
    "../artifacts/03_test.parquet",
    index=False
)

print("Saved:")
print("../artifacts/03_train.parquet")
print("../artifacts/03_validation.parquet")
print("../artifacts/03_test.parquet")
```
